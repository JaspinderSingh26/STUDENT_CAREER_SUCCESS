```python
# For each Major, calculate:
# Total Starting Salary
# Average Starting Salary
# Total Internships
# Average Employability Score

df.groupby("Major").agg(
    Total_Salary = ("Starting_Salary_USD", "sum"),
    Average_Salary = ("Starting_Salary_USD", "mean"),
    Total_Internships = ("Internships", "sum"),
    Average_Employability_Score = ("Employability_Score", "mean"))
```

```python
# For each University Year, calculate:
# Total Starting Salary
# Average Starting Salary
# Total Students
# Average Internships

df.groupby("University_Year").agg(
    Total_Salary = ("Starting_Salary_USD", "sum"),
    Average_Salary = ("Starting_Salary_USD", "mean"),
    Total_Students = ("Student_ID", "count"),
    Average_Internships = ("Internships", "mean"))
```

```python
# For each Major and Placement Mode, calculate:
# Total Starting Salary
# Total Students
# Average Employability Score

df.groupby(["Major", "Placement_Mode"]).agg(
    Total_Salary = ("Starting_Salary_USD", "sum"),
    Total_Students = ("Student_ID", "count"),
    Average_Employability_Score = ("Employability_Score", "mean"))
```

```python
# Which Major has the highest Total Starting Salary, and what is its Average Employability Score?
df.groupby("Major").agg(
    Total_Salary = ("Starting_Salary_USD", "sum"),
    Average_Employability_Score = ("Employability_Score", "mean")).sort_values("Total_Salary", ascending=False).head(1)
```

```python
# Find the Top 5 Career Fields based on Total Starting Salary, and show their Average Employability Score as well

df.groupby("Career_Field").agg(
    Total_Salary = ("Starting_Salary_USD", "sum"),
    Avg_Employability_Score = ("Employability_Score", "mean")).sort_values("Total_Salary", ascending=False).head()
```

```python
import matplotlib.pyplot as plt
```

# Total Starting Salary by Major

```python
# Adapted: replaced "Product_Category" with "Major", "Revenue" with "Starting_Salary_USD"

category_salary = df.groupby("Major")["Starting_Salary_USD"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(category_salary.index, category_salary.values, color="blue")

plt.xlabel("Major")
plt.ylabel("Total Starting Salary")
plt.title("Total Starting Salary by Major")

plt.xticks(rotation=45)
plt.show()
```

# Insight:  Computer Science generated the highest total starting salary among all majors, significantly outperforming the other majors.



#  Students by Payment Mode

```python
# Adapted: replaced "Payment_Mode" with "Placement_Mode", "Order_ID" with "Student_ID"

mode = df.groupby("Placement_Mode")["Student_ID"].count()

plt.figure(figsize=(7, 7))
plt.pie(
    mode.values,
    labels=mode.index,
    autopct="%1.1f%%"
)

plt.title("Students by Placement Mode")
plt.show()
```

# Insight: Campus placement is the most preferred placement mode among students.

# Major vs Avg_Employability_Score

```python
# Adapted: replaced "Product" with "Major", "Rating" with "Employability_Score"

category_salary = df.groupby("Major")["Starting_Salary_USD"].sum().sort_values(ascending=False)

colors = plt.cm.tab20.colors[:len(category_salary)]  # distinct color per bar

plt.figure(figsize=(8, 5))
plt.bar(category_salary.index, category_salary.values, color=colors)

plt.xlabel("Major")
plt.ylabel("Total Starting Salary")
plt.title("Total Starting Salary by Major")

plt.xticks(rotation=45)
plt.show()
```

# Insight: Artificial Intelligence has the highest average Employability Score among all majors.

# Starting Salary Trend by University Year

```python
# Adapted: no date column exists in this dataset, so using University_Year
# (which has a natural order) instead of Order_Date, and Starting_Salary_USD instead of Revenue

year_order = ["Freshman", "Sophomore", "Junior", "Senior"]
yearly_salary = df.groupby("University_Year")["Starting_Salary_USD"].sum().reindex(year_order)

plt.figure(figsize=(10, 5))
plt.plot(yearly_salary.index.astype(str), yearly_salary.values, marker="o")

plt.xlabel("University Year")
plt.ylabel("Total Starting Salary")
plt.title("Total Starting Salary by University Year")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Insight: Total starting salary rises steadily from Freshman through Junior year, then declines in Senior year

# Placement Status Distribution

```python
# Adapted: replaced "Order_Status" with "Placement_Status"

status = df["Placement_Status"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    status.values,
    labels=status.index,
    autopct="%1.1f%%",
    wedgeprops={"width": 0.4}
)

plt.title("Placement Status Distribution")
plt.show()
```

# Insight: The majority of students were successfully placed (78.08%), indicating a strong overall placement rate, while 21.92% remained unplaced.
